```python
import pandas as pd 
import seaborn as sns
```

```python
df = pd.read_csv("linear_data.csv")
df1 = pd.read_csv("nonlinear_data.csv")
```

```python
df.head()
```

```python
df.head()
```

```python
sns.scatterplot(x = df["x"], y = df["y"])
```

```python
sns.scatterplot(x = df1["x"], y = df1["y"])
```

```python
df.corr()
```

```python
df1.corr()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
lr = LinearRegression()
```

```python
linear_x= df[["x"]]
linear_y = df[["y"]]
non_l_x= df1[["x"]]
non_l_y = df1[["y"]]
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(linear_x, linear_y, test_size=.2, random_state=2)
```

```python
model1 = lr.fit(xtrain, ytrain)
model1.score(xtrain, ytrain)
```

```python
model1.score(xtest, ytest)
```

```python
# best fit line for linear data 
import matplotlib.pyplot as plt
pred = model1.predict(xtest)
```

```python
sns.scatterplot(x = df["x"], y=df["y"], color = "black")
plt.plot(xtest, pred)
plt.show()
```

# non linear data

```python
x_train, x_test, y_train, y_test = train_test_split(non_l_x, non_l_y, test_size=.2, random_state=2)
```

```python
model2 = lr.fit(x_train, y_train)
model2.score(x_train, y_train)
```

```python
model2.score(x_test, y_test)
```

```python
pred1 = model2.predict(x_test)
```

```python
sns.scatterplot(x = df1["x"], y=df1["y"], color = "black")
plt.plot(x_test, pred1)
plt.show()
```
